# Exercise for Graph II: Personalized Math Operations Agent

### Your Task:
Create a Graph where you pass in a single list of integers along with a name and an operation.
- If the operation is a `"+"`, add the elements.
- If the operation is a `"*"`, multiply the elements.
- All computation happens **within the same node**.

**Input:** `{"name": "Jack Sparrow", "values": [1, 2, 3, 4], "operation": "*"}`  
**Output:** `"Hi Jack Sparrow, your answer is: 24"`  
**Hint:** You need an if-statement in your node!

```python
# Install dependencies if needed:
# !pip install langgraph langchain-core
```

## 1. Define State and Node with `if` Statement

We define `MathState` containing:
- `name`: str
- `values`: list of integers
- `operation`: operator string (`"+"` or `"*"`)
- `result`: formatted output string

```python
import math
from typing import TypedDict, List
from langgraph.graph import StateGraph, START, END

# 1. Define the State
class MathState(TypedDict):
    name: str
    values: List[int]
    operation: str
    result: str

# 2. Define the single node with an if-statement to handle both operations
def calculate_node(state: MathState):
    name = state["name"]
    values = state["values"]
    operation = state["operation"]
    
    # Conditional logic within the same node
    if operation == "+":
        ans = sum(values)
    elif operation == "*":
        ans = math.prod(values)
    else:
        raise ValueError(f"Unsupported operation: {operation}")
    
    message = f"Hi {name}, your answer is: {ans}"
    return {"result": message}

# 3. Build and compile the Graph
workflow = StateGraph(MathState)
workflow.add_node("calculate", calculate_node)
workflow.add_edge(START, "calculate")
workflow.add_edge("calculate", END)

app = workflow.compile()
```

## 2. Test with the Exercise Input (Multiplication `*`)

```python
# Input from exercise
input_data = {
    "name": "Jack Sparrow",
    "values": [1, 2, 3, 4],
    "operation": "*"
}

output = app.invoke(input_data)

print("Final Output State:")
print(output)
print("\nAnswer Message:")
print(output["result"])
```

## 3. Test with Addition (`+`)

```python
# Addition test
add_input = {
    "name": "Jack Sparrow",
    "values": [1, 2, 3, 4],
    "operation": "+"
}

add_output = app.invoke(add_input)
print("Addition Result:")
print(add_output["result"])
```
